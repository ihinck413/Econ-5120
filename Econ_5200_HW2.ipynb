```python
import pandas as pd
import pandas_datareader.data as web
import datetime
import matplotlib.pyplot as plt

# Define the FRED series codes:
# PCEND: Personal Consumption Expenditures: Nondurable Goods (Billions of Dollars, Monthly)
# DNDGRG3M086SBEA: Personal consumption expenditures: Nondurable goods (implicit price deflator) (Index 2017=100, Monthly)
# POPTHM: Population (Thousands, Monthly)

series_dict = {
    'PCEND': 'Nominal_Nondurables_Billions',
    'DNDGRG3M086SBEA': 'Nondurables_Price_Index',
    'POPTHM': 'Population_Thousands'
}

start_date = datetime.datetime(1990, 1, 1)
end_date = datetime.date.today()

try:
    # Download the series from FRED
    df = web.DataReader(list(series_dict.keys()), 'fred', start_date, end_date)
    df.rename(columns=series_dict, inplace=True)

    # Display the first few rows to confirm successful download
    display(df.head())
except Exception as e:
    print(f"An error occurred while fetching data from FRED: {e}")
```

```python
# 1. Calculate Real Nondurable Consumption (in Billions of 2017 Dollars)
df['Real_Nondurables_Billions'] = df['Nominal_Nondurables_Billions'] / (df['Nondurables_Price_Index'] / 100.0)

# 2. Calculate Real Per-Capita Nondurable Consumption
df['Real_Per_Capita_Nondurables'] = (df['Real_Nondurables_Billions'] * 1000000) / df['Population_Thousands']

# Display the updated dataframe with calculations
display(df[['Nominal_Nondurables_Billions', 'Nondurables_Price_Index', 'Population_Thousands', 'Real_Per_Capita_Nondurables']].tail())
```

```python
import pandas as pd
import numpy as np

# 1. Load the monthly Fama-French data from the excel file
ff_monthly = pd.read_excel('/content/FF 3 Factor Quarterly.xls', skiprows=4)
ff_monthly.rename(columns={ff_monthly.columns[0]: 'Date'}, inplace=True)
ff_monthly = ff_monthly.dropna(subset=['Date'])
ff_monthly['Date'] = pd.to_datetime(ff_monthly['Date'], errors='coerce')
ff_monthly = ff_monthly.dropna(subset=['Date'])

# Clean and convert the factors to numeric (in decimal format for returns)
factor_cols = ['Mkt-RF', 'SMB', 'HML', 'RF']
for col in factor_cols:
    ff_monthly[col] = pd.to_numeric(ff_monthly[col], errors='coerce') / 100.0
ff_monthly = ff_monthly.dropna(subset=factor_cols)
ff_monthly.set_index('Date', inplace=True)

# 2. Compound monthly returns to get correct quarterly returns
# (1 + R_quarter) = (1 + R_m1) * (1 + R_m2) * (1 + R_m3)
# We group by calendar quarter ('QE') and calculate product of (1 + R)
ff_compounded = ff_monthly.apply(lambda x: x + 1.0)
ff_quarterly_comp = ff_compounded.resample('QE').prod() - 1.0

# Multiply back by 100 to keep it in percentage points
ff_quarterly_comp_pct = ff_quarterly_comp * 100.0

# 3. Align FRED quarterly data with the compounded quarterly Fama-French data
df_quarterly = df.resample('QE').mean() # keeping consumption as quarterly averages of monthly flows

merged_quarterly_comp = pd.merge(df_quarterly, ff_quarterly_comp_pct, left_index=True, right_index=True, how='inner')

# 4. Calculate log consumption growth
merged_quarterly_comp['Log_Real_Cons'] = np.log(merged_quarterly_comp['Real_Nondurables_Billions'])
merged_quarterly_comp['Log_Real_PerCapita_Cons'] = np.log(merged_quarterly_comp['Real_Per_Capita_Nondurables'])
merged_quarterly_comp['Log_Cons_Growth'] = merged_quarterly_comp['Log_Real_Cons'].diff()
merged_quarterly_comp['Log_PerCapita_Cons_Growth'] = merged_quarterly_comp['Log_Real_PerCapita_Cons'].diff()

# 5. Volatility (Annualized)
std_agg_q_comp = merged_quarterly_comp['Log_Cons_Growth'].std()
std_percapita_q_comp = merged_quarterly_comp['Log_PerCapita_Cons_Growth'].std()
std_agg_annual_comp = std_agg_q_comp * np.sqrt(4)
std_percapita_annual_comp = std_percapita_q_comp * np.sqrt(4)

# 6. Set up BOP and EOP growth rates
merged_quarterly_comp['EOP_Cons_Growth'] = merged_quarterly_comp['Log_Cons_Growth']
merged_quarterly_comp['EOP_PerCapita_Cons_Growth'] = merged_quarterly_comp['Log_PerCapita_Cons_Growth']
merged_quarterly_comp['BOP_Cons_Growth'] = merged_quarterly_comp['Log_Cons_Growth'].shift(-1)
merged_quarterly_comp['BOP_PerCapita_Cons_Growth'] = merged_quarterly_comp['Log_PerCapita_Cons_Growth'].shift(-1)

# 7. Correlations
eop_agg_corr_c = merged_quarterly_comp['EOP_Cons_Growth'].corr(merged_quarterly_comp['Mkt-RF'])
eop_percap_corr_c = merged_quarterly_comp['EOP_PerCapita_Cons_Growth'].corr(merged_quarterly_comp['Mkt-RF'])
bop_agg_corr_c = merged_quarterly_comp['BOP_Cons_Growth'].corr(merged_quarterly_comp['Mkt-RF'])
bop_percap_corr_c = merged_quarterly_comp['BOP_PerCapita_Cons_Growth'].corr(merged_quarterly_comp['Mkt-RF'])

# 8. Risk Aversion (Gamma) calculation with compounded returns
mkt_rf_dec_comp = merged_quarterly_comp['Mkt-RF'] / 100.0
mean_mkt_rf_q_c = mkt_rf_dec_comp.mean()
var_mkt_rf_q_c = mkt_rf_dec_comp.var()

mean_mkt_rf_annual_c = mean_mkt_rf_q_c * 4
var_mkt_rf_annual_c = var_mkt_rf_q_c * 4
std_mkt_rf_annual_c = np.sqrt(var_mkt_rf_annual_c)

lhs_c = mean_mkt_rf_annual_c + 0.5 * var_mkt_rf_annual_c

def compute_gamma_comp(std_c_annual, rho):
    cov_annual = std_c_annual * std_mkt_rf_annual_c * rho
    return lhs_c / cov_annual

gamma_eop_agg_c = compute_gamma_comp(std_agg_annual_comp, eop_agg_corr_c)
gamma_eop_percap_c = compute_gamma_comp(std_percapita_annual_comp, eop_percap_corr_c)
gamma_bop_agg_c = compute_gamma_comp(std_agg_annual_comp, bop_agg_corr_c)
gamma_bop_percap_c = compute_gamma_comp(std_percapita_annual_comp, bop_percap_corr_c)

print("--- NEW Compounded Quarterly Volatilities and Correlations ---")
print(f"Annualized Market Excess Return: {mean_mkt_rf_annual_c:.6f} | Market Vol: {std_mkt_rf_annual_c:.6f}")
print(f"EOP Aggregate Correlation: {eop_agg_corr_c:.6f} | BOP Aggregate Correlation: {bop_agg_corr_c:.6f}")
print(f"EOP Per-Capita Correlation: {eop_percap_corr_c:.6f} | BOP Per-Capita Correlation: {bop_percap_corr_c:.6f}")

print("\n--- NEW Implied CCAPM Risk Aversion (\u03b3) ---")
print(f"EOP Aggregate Consumption:     \u03b3 = {gamma_eop_agg_c:.2f}")
print(f"EOP Per-Capita Consumption:    \u03b3 = {gamma_eop_percap_c:.2f}")
print(f"BOP Aggregate Consumption:     \u03b3 = {gamma_bop_agg_c:.2f}")
print(f"BOP Per-Capita Consumption:    \u03b3 = {gamma_bop_percap_c:.2f}")
```
